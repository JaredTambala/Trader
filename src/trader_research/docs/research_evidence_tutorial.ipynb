# Research evidence identities

This notebook exercises the dependency-light identity, result, and canonical-reference vocabulary. It does not claim that an illustrative URI has been persisted.

```python
from trader_research.foundation import (
    parse_research_artifact_uri, research_artifact_uri,
    stable_research_id, success_result,
)

scope = {"asset_class": "stocks", "symbols": ["AAPL", "MSFT"], "timeframe": "1Hour"}
manifest_id = stable_research_id("dataset_manifest", scope)
assert manifest_id == stable_research_id("dataset_manifest", scope)
uri = research_artifact_uri("dataset_manifest", manifest_id)
assert parse_research_artifact_uri(uri) == ("dataset_manifest", manifest_id)
```

```python
outcome = success_result(
    command="tutorial_data_scope",
    data={"scope": scope},
    artifacts={"dataset_manifest": uri},
    warnings=("illustrative reference only",),
)
payload = outcome.to_dict()
assert payload["ok"] is True
assert payload["artifacts"]["dataset_manifest"] == uri
assert payload["warnings"] == ["illustrative reference only"]
```
